## About Dataset
Context

This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

Content
5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset.


Columns

- asin - ID of the product, like B000FA64PK
- helpful - helpfulness rating of the review - example: 2/3.
- overall - rating of the product.
- reviewText - text of the review (heading).
- reviewTime - time of the review (raw).
- reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
- reviewerName - name of the reviewer.
- summary - summary of the review (description).
- unixReviewTime - unix timestamp.

## Preprocessing and Cleaning 

```python
# Normalization
# Positive review is 1 and negative review is 0 ( 3 above, positive. Below 3, Negative)
df["rating"] = data["rating"].apply(lambda x: 0 if x < 3 else 1)
```

```python
# Convert all review text to lowercase for consistent text processing
df["reviewText"].str.lower()
```

```python
# 1. Lower All the cases
df["reviewText"] = df["reviewText"].str.lower()
```

```python
# # Import the 're' module for working with regular expressions
# Import the Natural Language Toolkit (NLTK) library for processing and analysing human language
# Import the stopwords corpus from NLTK to remove the common words during text processing
# Download the stopwords datasets required by nltk

import re
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python

# Import BeautifulSoup from the bs4 library.
# to parse HTML or XML documents and extract data.
from bs4 import BeautifulSoup
```

```python
# Make a copy to avoid SettingWithCopyWarning
df = df.copy()

# Load stopwords once (faster)
stop_words = set(stopwords.words('english'))

# Handle missing values
df['reviewText'] = df['reviewText'].fillna('').astype(str)

# 1. Remove URLs
df['reviewText'] = df['reviewText'].apply(
    lambda x: re.sub(r'https?://\S+|www\.\S+', '', x))

# 2. Remove HTML tags
df['reviewText'] = df['reviewText'].apply(
    lambda x: BeautifulSoup(x, 'lxml').get_text())

# 3. Remove special characters
df['reviewText'] = df['reviewText'].apply(
    lambda x: re.sub(r'[^a-zA-Z0-9\s-]', '', x))

# 4. Convert to lowercase
df['reviewText'] = df['reviewText'].apply(lambda x: x.lower())

# 5. Remove stopwords
df['reviewText'] = df['reviewText'].apply(lambda x: " ".join(
    word for word in x.split()if word not in stop_words))

# 6. Remove extra spaces
df['reviewText'] = df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
# Display cleaned Text
df.head()
```

```python
# Lemmatizer

# Import WordNetLemmatizer from NLTK.
from nltk.stem import WordNetLemmatizer
```

```python
# Create an instance of the lemmatizer.
lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    # Split the input text into individual words.
    # Apply lemmatization to each word using the lemmatizer.
    # Join the lemmatized words back into a single string with spaces.
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python

# Apply the lemmatize_words() function to each review
# in the 'reviewText' column using the apply() method.
# The lambda function passes each review (x) to the
# lemmatize_words() function.
# Store the lemmatized reviews back in the same column.
df['reviewText'] = df['reviewText'].apply(
    lambda x: lemmatize_words(x)
)
```

```python
df.head()
```

```python
# import the train_test_split function from the sklearn.model_selection module.
# split the dataset into training and testing sets using the train_test_split() function.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['reviewText'], df['rating'],
                                                    test_size=0.20, random_state=42, stratify=df['rating'])
```

```python

# Import CountVectorizer to convert text into numerical features
# using the Bag of Words (BoW) technique.
from sklearn.feature_extraction.text import CountVectorizer

# Create a Bag of Words vectorizer.
# To Remove common English stop words and keep at most 10,000 features.
bow = CountVectorizer(
    stop_words="english",
    max_features=10000
)

# Learn the vocabulary from the training reviews and convert
# them into a numerical document-term matrix.
X_train_bow = bow.fit_transform(X_train).toarray()

# Convert the test reviews using the same vocabulary learned
# from the training data, without learning new vocabulary.
X_test_bow = bow.transform(X_test).toarray()
```

```python
# Import TfidfVectorizer to convert text into numerical features
# based on the importance of words in each document.
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the TF-IDF vectorizer.
# Remove common English stop words and limit the vocabulary
# to a maximum of 10,000 features.
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=10000
)

# Learn the vocabulary and calculate TF-IDF features
# from the training reviews.
X_train_tfidf = tfidf.fit_transform(X_train).toarray()

# Transform the test reviews using the vocabulary and
# IDF values learned from the training data.
X_test_tfidf = tfidf.transform(X_test).toarray()
```

```python
# Import the machine learning models
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

# Evaluation Metrics
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import confusion_matrix
```

```python
models = {
    "Naive Bayes": MultinomialNB(),
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight="balanced"),
    "Linear SVM": LinearSVC(),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),

}
```

```python
def train_models(models, X_train, X_test, y_train, y_test):

    # store the accuracy result for each model
    results = {}

    # train and evaluate each models in a dictionary format
    for name, model in models.items():

        print("\n")
        print("==============================")
        print(name)
        print("==============================")

        # model training
        model.fit(X_train, y_train)

        # Make predictions
        y_pred = model.predict(X_test)

        # accuracy score
        accuracy = accuracy_score(y_test, y_pred)

        results[name] = accuracy
        print("Accuracy:", accuracy)

        # Detailed performance report using the classification report output
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred)
              )

        # Confusion matrix
        print("\nConfusion Matrix:")
        print(confusion_matrix(y_test, y_pred)
              )

    # Return all model evaluation results
    return results
```

```python
# Train and evaluate the machine learning models
# using Bag of Words (BoW) features.
bow_results = train_models(
    models,
    X_train_bow,
    X_test_bow,
    y_train,
    y_test
)
```

 Class 0: Precision is 73%, meaning that 73% of reviews predicted as class 0 were actually class 0. Recall is 80%, the model correctly identified 80% of all actual class 0 reviews.
 Class 1: Precision is 90%, and recall is 85%. The model performs better at identifying class 1 than class 0 overall.
 The model is better at predicting class 1, based on its higher precision and F1-score.



 So, based on this confusion matrix, Logistic Regression has 70.75% accuracy, which is lower than Naive Bayes at 83.46%.

 Naive Bayes performs better than Logistic Regression on the supplied test data. 
 It makes 305 more correct predictions and has about 12.71 percentage points higher accuracy.

```python

# Train and evaluate the machine learning models
# using TF-IDF numerical features.
tfidf_results = train_models(
    models,
    X_train_tfidf,
    X_test_tfidf,
    y_train,
    y_test
)
```

```python
# Tokenize the X_trains
X_train_tokens = [review.split() for review in X_train]
X_test_tokens = [review.split() for review in X_test]
```

```python
# Generate the word2vec model and implement it on the data
from gensim.models import Word2Vec

word2vec_model = Word2Vec(
    sentences=X_train_tokens,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    epochs=20
)
```

```python
# Transform text reviews into numerical features using Word2Vec embeddings.
```

```python
import numpy as np
"""
    Convert a single review (list of words) into one Word2Vec vector
    
"""

# Get the Word2Vec embedding for each word that exists in the trained vocabulary


def review_vector(review):
    vectors = [
        word2vec_model.wv[word]
        for word in review
        if word in word2vec_model.wv
    ]

    # return a vector of zeros (100 dimensions)
    # If no words from the review are found in the Word2Vec vocabulary
    if len(vectors) == 0:
        return np.zeros(100)

    return np.mean(vectors, axis=0)


X_train_word2vec = np.array([review_vector(r) for r in X_train_tokens])
X_test_word2vec = np.array([review_vector(r) for r in X_test_tokens])
```
